```python
df = df.drop(columns=['id', 'Unnamed: 32'])

df.shape
```

```python
df.head()
```

```python
df['diagnosis'] = df['diagnosis'].map({
    'B': 0,
    'M': 1
})

df['diagnosis'].value_counts()
```

```python
X = df.drop(columns=['diagnosis'])
y = df['diagnosis']

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split( X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
   
)
   

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.metrics import accuracy_score
from sklearn.ensemble import GradientBoostingClassifier

gb_model = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)

gb_model.fit(X_train, y_train)

gb_pred = gb_model.predict(X_test)

gb_prob = gb_model.predict_proba(X_test)

gb_benign_prob = gb_prob[:, 0]
gb_malignant_prob = gb_prob[:, 1]

print("Gradient Boosting Accuracy:", accuracy_score(y_test, gb_pred))
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score, roc_auc_score

print("Precision:", precision_score(y_test, gb_pred))
print("Recall:", recall_score(y_test, gb_pred))
print("F1-Score:", f1_score(y_test, gb_pred))
print("ROC-AUC:", roc_auc_score(y_test, gb_malignant_prob))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, gb_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=['Benign', 'Malignant']
)

disp.plot()
plt.title("Gradient Boosting - Confusion Matrix")
plt.show()
```

```python
from sklearn.metrics import classification_report

print(classification_report(
    y_test,
    gb_pred,
    target_names=['Benign', 'Malignant']
))
```

```python
from sklearn.metrics import roc_curve
import matplotlib.pyplot as plt

fpr, tpr, thresholds = roc_curve(y_test, gb_malignant_prob)

plt.plot(fpr, tpr)
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Gradient Boosting - ROC Curve")
plt.show()
```

```python
import matplotlib.pyplot as plt

comparison_table.set_index("Model").plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Performance Comparison of All Models")
plt.xlabel("Models")
plt.ylabel("Score")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()

plt.show()
```

```python
from sklearn.metrics import roc_curve
import matplotlib.pyplot as plt

# Calculate ROC curves
fpr_gb, tpr_gb, _ = roc_curve(y_test, gb_malignant_prob)
fpr_ada, tpr_ada, _ = roc_curve(y_test, ada_malignant_prob)
fpr_xgb, tpr_xgb, _ = roc_curve(y_test, xgb_malignant_prob)
fpr_cat, tpr_cat, _ = roc_curve(y_test, cat_malignant_prob)
fpr_rf, tpr_rf, _ = roc_curve(y_test, rf_malignant_prob)
fpr_lgbm, tpr_lgbm, _ = roc_curve(y_test, lgbm_malignant_prob)

# Plot ROC curves
plt.figure(figsize=(10, 7))

plt.plot(fpr_gb, tpr_gb, label="Gradient Boosting (AUC = 0.9947)")
plt.plot(fpr_ada, tpr_ada, label="AdaBoost (AUC = 0.9944)")
plt.plot(fpr_xgb, tpr_xgb, label="XGBoost (AUC = 0.9944)")
plt.plot(fpr_cat, tpr_cat, label="CatBoost (AUC = 0.9987)")
plt.plot(fpr_rf, tpr_rf, label="Random Forest (AUC = 0.9940)")
plt.plot(fpr_lgbm, tpr_lgbm, label="LGBM (AUC = 0.9960)")

# Random classifier line
plt.plot([0, 1], [0, 1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC Curve Comparison of All Models")

plt.legend()

plt.grid()

plt.tight_layout()

plt.show()
```
